#### STFT

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from obspy import read
from scipy.signal import stft
```

```python
# -------- CONFIGURATION --------
input_folder = "C:/work/all_paper/papers/p3/Final submission/sr/final_editted/plot_EQ/"       
output_folder = "C:/work/all_paper/papers/p3/Final submission/sr/final_editted/plot_EQ"       

os.makedirs(output_folder, exist_ok=True)
```

```python
# -------- LOOP THROUGH FILES --------
for filename in os.listdir(input_folder):
    if filename.endswith(".mseed"):
        file_path = os.path.join(input_folder, filename)
        try:

            st = read(file_path)
            tr = st[0]
            data = tr.data
            sampling_rate = tr.stats.sampling_rate
            times = np.arange(len(data)) / sampling_rate

            data = (data - np.mean(data)) / np.std(data)

            f, t, Zxx = stft(data, fs=sampling_rate, window='hann', nperseg=256, noverlap=128)
            power = np.abs(Zxx)

            plt.figure(figsize=(10, 6))
            plt.pcolormesh(t, f, power, shading='gouraud', cmap='viridis')
            plt.colorbar(label='Amplitude')
            plt.xlabel('Time (s)')
            plt.ylabel('Frequency (Hz)')
            plt.title(f'STFT: {tr.id} | {tr.stats.starttime.isoformat()} — {tr.stats.endtime.isoformat()}')
            plt.ylim(0, 30)
            plt.tight_layout()

            save_path = os.path.join(output_folder, f"{os.path.splitext(filename)[0]}_stft1.png")
            plt.savefig(save_path, dpi=450)
            plt.close()
            print(f"Saved: {save_path}")

        except Exception as e:
            print(f"Error processing {filename}: {e}")
```
